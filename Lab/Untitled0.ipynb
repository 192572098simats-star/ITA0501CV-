```python
import cv2
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread("cat.jpg")

if img is None:
    print("Image not found")
else:
    # Apply Gaussian Blur
    blur = cv2.GaussianBlur(img, (15, 15), 0)

    # Display input and output
    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(cv2.cvtColor(blur, cv2.COLOR_BGR2RGB))
    plt.title("Blurred Image")
    plt.axis("off")

    plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread("cat.jpg")

if img is None:
    print("Image not found")
else:
    # Convert to grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Detect edges using Canny
    edges = cv2.Canny(gray, 100, 200)

    # Display original image
    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title("Original Image")
    plt.axis("off")

    # Display outline image
    plt.subplot(1, 2, 2)
    plt.imshow(edges, cmap="gray")
    plt.title("Canny Edge Detection")
    plt.axis("off")

    plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

# Read the image
img = cv2.imread("cat.jpg")

if img is None:
    print("Image not found")
else:
    # Convert to grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Create kernel
    kernel = np.ones((5, 5), np.uint8)

    # Dilate the image
    dilated = cv2.dilate(gray, kernel, iterations=1)

    # Display original image
    plt.subplot(1, 2, 1)
    plt.imshow(gray, cmap="gray")
    plt.title("Original Grayscale")
    plt.axis("off")

    # Display dilated image
    plt.subplot(1, 2, 2)
    plt.imshow(dilated, cmap="gray")
    plt.title("Dilated Image")
    plt.axis("off")

    plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

# Read the image
img = cv2.imread("cat.jpg")

if img is None:
    print("Image not found")
else:
    # Convert to grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Create kernel
    kernel = np.ones((5, 5), np.uint8)

    # Erode the image
    eroded = cv2.erode(gray, kernel, iterations=1)

    # Display original image
    plt.subplot(1, 2, 1)
    plt.imshow(gray, cmap="gray")
    plt.title("Original Image")
    plt.axis("off")

    # Display eroded image
    plt.subplot(1, 2, 2)
    plt.imshow(eroded, cmap="gray")
    plt.title("Eroded Image")
    plt.axis("off")

    plt.show()
```

```python
import cv2
from google.colab.patches import cv2_imshow

img = cv2.imread("cat.jpg")

bigger = cv2.resize(img, None, fx=2, fy=2)
smaller = cv2.resize(img, None, fx=0.5, fy=0.5)

print("Original Image")
cv2_imshow(img)

print("Bigger Image")
cv2_imshow(bigger)

print("Smaller Image")
cv2_imshow(smaller)
```

```python
import cv2
from google.colab.patches import cv2_imshow

img = cv2.imread("cat.jpg")

clockwise = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)
counter_clockwise = cv2.rotate(img, cv2.ROTATE_90_COUNTERCLOCKWISE)

print("Original Image")
cv2_imshow(img)

print("Clockwise Rotation")
cv2_imshow(clockwise)

print("Counter Clockwise Rotation")
cv2_imshow(counter_clockwise)
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("cat.jpg")

rows, cols = img.shape[:2]

# Move image 100 pixels right and 50 pixels down
matrix = np.float32([[1, 0, 100],
                     [0, 1, 50]])

moved = cv2.warpAffine(img, matrix, (cols, rows))

print("Original Image")
cv2_imshow(img)

print("Moved Image")
cv2_imshow(moved)
```
